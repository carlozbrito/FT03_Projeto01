```python
def sidra_json_to_df(data):
    """Flatten an IBGE SIDRA (servicodados.ibge.gov.br) response into one row per
    variable x classification combination x locality x period."""
    rows = []
    for var in data:
        for res in var["resultados"]:
            # One column per classification (e.g. 'Situação do domicílio' -> 'Urbana')
            classif = {c["nome"]: ", ".join(c["categoria"].values()) for c in res["classificacoes"]}
            for s in res["series"]:
                loc = s["localidade"]
                for periodo, valor in s["serie"].items():
                    rows.append({
                        "variavel_id": var["id"],
                        "variavel": var["variavel"],
                        "unidade": var["unidade"],
                        **classif,
                        "localidade_id": loc["id"],
                        "localidade": loc["nome"],
                        "nivel": loc["nivel"]["nome"],
                        "periodo": periodo,
                        "valor": valor,
                    })
    df = pd.DataFrame(rows)
    # IBGE uses symbols like '-', '...', 'X' for missing/suppressed values -> NaN
    df["valor"] = pd.to_numeric(df["valor"], errors="coerce")
    return df
 
```

```python
def ibge_resultados_to_df(resultados):
    """Flatten a list of IBGE 'resultados' blocks: one row per classification combo / location / period.
    Each classification (e.g. 'Tipos de despesa') becomes its own column."""
    rows = []
    for res in resultados:
        classif = {c['nome']: ', '.join(c['categoria'].values()) for c in res['classificacoes']}
        for s in res['series']:
            loc = s['localidade']
            for periodo, valor in s['serie'].items():
                rows.append({
                    **classif,
                    'nivel': loc['nivel']['nome'],
                    'localidade_id': loc['id'],
                    'localidade': loc['nome'],
                    'periodo': int(periodo),
                    'valor': pd.to_numeric(valor, errors='coerce'),  # '-', '...', 'X' -> NaN
                })
    return pd.DataFrame(rows)
```

## Qtde viagens (mil)

```python
import requests
import pprint

link = "https://servicodados.ibge.gov.br/api/v3/agregados/7204/periodos/2024/variaveis/10390?localidades=N3[all]&classificacao=892[47976,47977]"
requisicao = requests.get(link)
informacoes_viagens = requisicao.json()
```

```python
informacoes_viagens 
```

```python
rows = []
for var in informacoes_viagens:
    for res in var["resultados"]:
        cls = res["classificacoes"][0]
        cat_id, cat_nome = next(iter(cls["categoria"].items()))
        for s in res["series"]:
            loc = s["localidade"]
            for ano, valor in s["serie"].items():
                rows.append({
                    "variavel_id": var["id"],
                    "variavel": var["variavel"],
                    "unidade": var["unidade"],
                    "classificacao": cls["nome"],
                    "categoria_id": cat_id,
                    "finalidade": cat_nome,
                    "uf_id": loc["id"],
                    "uf": loc["nome"],
                    "nivel": loc["nivel"]["nome"],
                    "ano": int(ano),
                    "viagens_mil": pd.to_numeric(valor, errors="coerce"),  # IBGE uses '-'/'...' for missing
                })
 
df = pd.DataFrame(rows)
```
